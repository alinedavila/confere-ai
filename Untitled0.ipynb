Projeto de Previsão e Modelagem

```python
# Projeto de Previsão e Modelagem

situacoes = {
    "Altura média dos estudantes daqui a um ano": {
        "dificuldade": "Fácil",
        "estabilidade": "Estável",
        "informacao": "Altura atual dos estudantes e histórico de crescimento por idade."
    },

    "Tempo médio de deslocamento até a escola": {
        "dificuldade": "Fácil",
        "estabilidade": "Relativamente estável",
        "informacao": "Tempo de deslocamento registrado em diferentes dias e horários."
    },

    "Vídeo mais assistido da internet na próxima semana": {
        "dificuldade": "Difícil",
        "estabilidade": "Muito volátil",
        "informacao": "Visualizações, curtidas e compartilhamentos dos vídeos que estão em alta."
    },

    "Preço de uma moeda estrangeira amanhã": {
        "dificuldade": "Difícil",
        "estabilidade": "Volátil",
        "informacao": "Histórico da cotação, juros, inflação e acontecimentos econômicos."
    }
}

for situacao, dados in situacoes.items():
    print("Situação:", situacao)
    print("Dificuldade:", dados["dificuldade"])
    print("Estabilidade:", dados["estabilidade"])
    print("Informação útil:", dados["informacao"])
    print("-" * 60)
```

## Análise

As situações consideradas mais fáceis de modelar são a previsão da altura média dos estudantes daqui a um ano e o tempo médio de deslocamento até a escola. A altura tende a variar de maneira gradual, enquanto o tempo de deslocamento pode apresentar padrões dependendo do horário e do trânsito.

As situações consideradas mais difíceis são a previsão do vídeo mais assistido da internet e o preço de uma moeda estrangeira amanhã. As duas são mais voláteis e podem sofrer mudanças por acontecimentos inesperados. No caso dos vídeos, tendências e acontecimentos nas redes sociais podem alterar rapidamente as visualizações. Já a cotação de uma moeda pode mudar devido a fatores econômicos e políticos.
